```python
import joblib
import numpy as np

# Load files
model = joblib.load("lightgbm.pkl")
scaler = joblib.load("vitals_scaler.pkl")
encoder = joblib.load("risk_label_encoder.pkl")

# ==========================
# TEST FEATURES
# Order:
# HR, RR, Temp, SpO2, SysBP, DiaBP,
# Age, Gender, Weight, Height,
# HRV, PulsePressure, BMI, MAP
# ==========================

features = [[
    70,      # Heart Rate
    18,      # Respiratory Rate
    36.8,    # Body Temperature
    98,      # Oxygen Saturation
    120,     # Systolic BP
    80,      # Diastolic BP
    30,      # Age
    1,       # Gender (Male=1, Female=0)
    70,      # Weight
    1.75,    # Height
    0.05,    # HRV
    40,      # Pulse Pressure
    22.86,   # BMI
    93.33    # MAP
]]

# Scale
X = scaler.transform(features)

# Predict
pred = model.predict(X)[0]
probs = model.predict_proba(X)[0]

# Decode prediction
predicted_label = encoder.inverse_transform([pred])[0]

print("Classes:", encoder.classes_)
print("Raw Prediction:", pred)
print("Decoded Prediction:", predicted_label)

print("\nProbabilities:")
for i, cls in enumerate(encoder.classes_):
    print(f"{cls}: {probs[i]:.6f}")
```

```python
test_cases = [
    [
        70,18,36.6,98,120,80,
        30,1,70,1.75,
        0.05,40,22.86,93.33
    ],

    [
        120,30,40,85,180,120,
        75,1,95,1.65,
        0.50,60,35,140
    ]
]

X = scaler.transform(test_cases)

preds = model.predict(X)
probs = model.predict_proba(X)

for i in range(len(test_cases)):
    label = encoder.inverse_transform([preds[i]])[0]

    print(f"\n===== CASE {i+1} =====")
    print("Prediction:", label)

    for j, cls in enumerate(encoder.classes_):
        print(f"{cls}: {probs[i][j]:.6f}")
```
